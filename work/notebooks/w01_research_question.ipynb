## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
```

This work improves the decision a content strategist makes about which pages to review
and refresh first, out of potentially thousands of pages. The action: they open the top of
a ranked list and manually review/update those pages. The cost of a wrong call: flagging a
page that didn't actually need attention wastes reviewer time (false positive); missing a
page that's genuinely declining means real traffic loss that compounds the longer it goes
unnoticed (false negative, the costlier mistake). A plain if-statement rule isn't quite
enough because staleness, visibility, and decline interact in ways that are hard to hand-tune
precisely -- which is why a learned model is worth testing against the baseline rule.
